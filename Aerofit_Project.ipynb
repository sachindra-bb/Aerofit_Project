```python
# Load the dataset
url = "https://d2beiqkhq929f0.cloudfront.net/public_assets/assets/000/001/125/original/aerofit_treadmill.csv?1639992749"
```

```python
#create the datadframe
import pandas as pd
aerofit_data = pd.read_csv(url)
# Observing basic structure and metrics
print("Shape of the dataset:")
print(aerofit_data.shape)
print("\n")
print("Data Types of each column:")
print(aerofit_data.dtypes)
print("\n")
```

```python
# Convert relevant columns to 'category'
categorical_columns = ['Product', 'Gender', 'MaritalStatus']
for col in categorical_columns:
  aerofit_data[col] = aerofit_data[col].astype('category')
  print("Data Types after conversion:")
  print(aerofit_data.dtypes)
  print("\n")
```

```python
# Statistical summary of the numerical columns
print("Statistical Summary:")
print(aerofit_data.describe())
# Count the number of records for each category
product_counts = aerofit_data ['Product'].value_counts()
gender_counts = aerofit_data ['Gender'].value_counts()
marital_status_counts = aerofit_data ['MaritalStatus'].value_counts()
```

```python
# Display the counts
print("Counts by Product:")
print(product_counts)
print()
print("Counts by Gender:")
print(gender_counts)
print()
print("Counts by Marital Status:")
print(marital_status_counts)
```

```python
# Value counts for each categorical column
product_value_counts = aerofit_data['Product'].value_counts()
gender_value_counts = aerofit_data['Gender'].value_counts()
marital_status_value_counts = aerofit_data['MaritalStatus'].value_counts()
# Displaying value counts
print("Value counts for Product:")
print(product_value_counts)
print()
print("Value counts for Gender:")
print(gender_value_counts)
print()
print("Value counts for MaritalStatus:")
print(marital_status_value_counts)
print()
# Count unique attributes for all columns
unique_attributes = aerofit_data.nunique()
# Display the number of unique attributes for each column
print("Unique attributes in each column:")
print(unique_attributes)
```

```python
#Visual Analysis Univariate and Bivariate
import seaborn as sns
import matplotlib.pyplot as plt
# Univariate Analysis for Continuous Variables (e.g., Age, Education, Income)
# Distplot for Age (can change continuous columns as per your analysis)
plt.figure(figsize=(10, 6))
sns.histplot(aerofit_data['Age'], kde=True, bins=20, color='skyblue')
plt.title('Distribution of Age')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.show()
# Countplot for Product - Checking distribution of product across customers
plt.figure(figsize=(8, 5))
sns.countplot(x='Product', data=aerofit_data, palette='viridis')
plt.title('Count of Products Purchased')
plt.xlabel('Product')
plt.ylabel('Count')
plt.show()
# Histogram for Income
plt.figure(figsize=(10, 6))
sns.histplot(aerofit_data['Income'], kde=True, bins=20, color='orange')
plt.title('Distribution of Income')
plt.xlabel('Income')
plt.ylabel('Frequency')
plt.show()
```

```python
# Univariate Boxplot for Categorical Variables (e.g., Gender, MaritalStatus)
plt.figure(figsize=(8, 6))
sns.boxplot(x='Gender', y='Income', data=aerofit_data, palette='Set2')
plt.title('Income Distribution by Gender')
plt.xlabel('Gender')
plt.ylabel('Income')
plt.show()
# Boxplot for Marital Status and Fitness Level
plt.figure(figsize=(8, 6))
sns.boxplot(x='MaritalStatus', y='Fitness', data=aerofit_data, palette='Set3')
plt.title('Fitness Level by Marital Status')
plt.xlabel('Marital Status')
plt.ylabel('Fitness')
plt.show()
```

```python
#showing correlations and heat maps
# Bivariate Analysis - Correlation Heatmap for Continuous Variables
# Selecting continuous columns
continuous_columns = ['Age', 'Education', 'Income', 'Fitness', 'Miles']
# Compute correlation matrix
corr_matrix = aerofit_data[continuous_columns].corr()
# Plot heatmap
plt.figure(figsize=(10, 8))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt='.2f', linewidths=0.5)
plt.title('Correlation Heatmap for Continuous Variables')
plt.show()
# Add blank line for better understanding
print("\n" + "-"*50 + "\n")
# Pairplot for Continuous Variables to explore pairwise relationships
sns.pairplot(aerofit_data[continuous_columns], diag_kind='kde', plot_kws={'alpha': 0.5})
plt.suptitle('Pairplot of Continuous Variables', y=1.02)
plt.show()
# Add blank line for better understanding
print("\n" + "-"*50 + "\n")
```

```python
#Checking for Missing Values
missing_values = aerofit_data.isnull().sum()
print("Missing Values in Each Column:")
print(missing_values)
print("\n" + "-"*50 + "\n")
# If we have any missing values, we might fill them with median (for numerical) or mode (for categorical), or drop rows with missing values.
# Example: Filling missing values for numerical columns with median and categorical columns with mode
# Fill missing numerical columns with median
numerical_columns = aerofit_data.select_dtypes(include=['int64', 'float64']).columns
for col in numerical_columns:
  median_value = aerofit_data[col].median()
  aerofit_data[col].fillna(median_value, inplace=True)
# Fill missing categorical columns with mode
categorical_columns = aerofit_data.select_dtypes(include=['category', 'object']).columns
for col in categorical_columns:
  mode_value = aerofit_data[col].mode()[0]
  aerofit_data[col].fillna(mode_value, inplace=True)
```

```python
# Outlier Detection Using IQR for Numerical Columns
# Using IQR (Interquartile Range) to detect outliers
for col in numerical_columns:
  Q1 = aerofit_data[col].quantile(0.25)
  Q3 = aerofit_data[col].quantile(0.75)
  IQR = Q3 - Q1
  lower_bound = Q1 - 1.5 * IQR
  upper_bound = Q3 + 1.5 * IQR
  outliers = aerofit_data[(aerofit_data[col] < lower_bound) | (aerofit_data[col] > upper_bound)]
  print(f"Outliers in {col}:")
  print(f"Lower Bound: {lower_bound}, Upper Bound: {upper_bound}")
  print(f"Number of Outliers: {len(outliers)}\n")
  print("-"*50)
# visualizing outliers using boxplots
import seaborn as sns
import matplotlib.pyplot as plt
# Visualizing outliers with boxplots
for col in numerical_columns:
  plt.figure(figsize=(8, 5))
  sns.boxplot(x=aerofit_data[col], color='lightblue')
  plt.title(f"Boxplot of {col}")
  plt.xlabel(col)
  plt.show()
```
